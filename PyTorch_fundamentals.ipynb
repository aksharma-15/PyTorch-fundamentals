### Matrix multiplication

Two main ways of performing multiplication in neural networks and deep learning.
1. Element-wise multiplication
2. Matrix to matrix multiplication

There are two rules that performing matrix multiplication needs to satisfy:
1. The **inner dimensions** must match:
  * `(3, 2) @ (3, 2)` won't work
  * `(2, 3) @ (3, 2)` will work
  * `(3, 2) @ (2, 3)` will work

2. The resulting matrix has the shape of the **outer dimensions**:
  * `(2, 3) @ (3, 2)` -> `(2, 2)`
  * `(3, 2) @ (2, 3)` -> `(3, 3)`

```python
# example - this won't work
# torch.matmul(torch.rand(3, 2), torch.rand(3, 2))
```

```python
# Element-wise multiplication
tensor = torch.Tensor([1, 2, 3])

print(tensor, '*', tensor)
print(f'Equals: {tensor * tensor}')
```

```python
# Matrix multiplication

%%time
torch.matmul(tensor, tensor) # can also use torch.mm()
```

```python
# Similar to torch.matmul()

%%time
tensor @ tensor
```

```python
# Matrix multiplication by hand

%%time
value  = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]

print(value)
```

* Traditional python for loop is almost 10x slower than Pytorch's vectorized functions

### One of the most common errors in deep learning is shape errors.

```python
# Shapes for matrix multiplication
tensor_A = torch.Tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

# torch.mm(tensor_A, tensor_B) # torch.mm is the same as torch.matmul (it's an alias for writing less code)
tensor_A.shape, tensor_B.shape # won't work as the inner dimensions must match
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors

A **transpose** switches the axes or dimensions of a given tensor.

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
# The matrix multiplication operation works when tensor_B is transposed
print(f'Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}')
print(f'New shapes: tensor_A: {tensor_A.shape} (same shape as above), tensor_B.T = {tensor_B.T.shape}')
print(f'Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} -> inner dimensions must match')
print(f'Output:\n')
output = torch.matmul(tensor_A, tensor_B.T.to(tensor_A.dtype)) # without using .to(tensor_A.dtype) we will get a traceback in the latest version of pytorch
print(output)
print(f'\nOutput shape: {output.shape}')
```

## Reshaping, stacking, squeezing and unsqueezing tensors.
* Reshaping - reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
# Let's create a tensor
x = torch.arange(1., 10.)
x, x.shape, x.size(), x.dtype
```

```python
# Add an extra dimension # total elements should equal to the original tensor
x_reshaped = x.reshape(3, 3)
x_reshaped, x_reshaped.shape
```

```python
# or it can be a column tensor
x_col = x.reshape(9,1)
x_col, x_col.shape
```

```python
# Change the view
z = x.view(1, 9)
z, z.shape
```

```python
# Changing z changes x (because a view of a tensor shares the memory as the original tensor)
z[:, 0] = 5
z, x
```

```python
import numpy as np # we can also use numpy to check if memory is being shared bw the 2 tensors
np.shares_memory(z, x)
```

```python
# Stack tensor on top of each other
x_stacked = torch.stack([x, x, x, x], dim = 0) # stacked vertically
x_stacked
```

```python
x_stacked = torch.stack([x, x, x, x], dim = 1) # stacked horizontally
x_stacked
```

```python
# torch.squeeze - removes all single dimensions from a target tensor
x_reshaped, x_reshaped.shape, x_reshaped.ndim
```

```python
# squeeze
x_reshaped.squeeze()
```

```python
# torch.unsqueeze() - adds a single dimension to a target dimension at a specific dimension
x_reshaped.unsqueeze(dim = 0)
```

```python
# torch.permute - rearranges the dimensions of a target tensor in a specific order
x_original = torch.rand(size = (224, 224, 3)) # [height, width, colour_channels]

# Permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->, 1->2, 2->0

print(f'Previous shape: {x_original.shape}')
print(f'New shape: {x_permuted.shape}') # [colour_channels, height, width]
```

```python
x_original[0, 0, 0] = 728218
x_original[0, 0, 0], x_permuted[0, 0, 0]
```

## PyTorch tensors and NumPy

NumPy is a popular scientific python numerical computing library and because of this , PyTorch has functionality to interact with it.

* Data in NumPy, want in PyTorch tensor -> `torch.from_numpy(ndarray)`
* PyTorch tensor -> NumPy -> `torch.Tensor.numpy()`

```python
# NumPy array to tensor
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
# tensor = torch.from_numpy(array).type(torch.float32) to get the tensor in pytorch's default dtype
array, tensor # default dtype will be float64 as it is numpy's default dtype
```

```python
# change the value of array, what will this do to `tensor`?
array = array + 1
array, tensor # it doesn't alter the tensor
```

```python
# Tensor to numpy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor # changes the array's dtype to float32 when converted into a tensor
```

```python
# Change the tensor, what happens to 'numpy_tensor'
tensor = tensor + 1
tensor, numpy_tensor # don't share memory
```

## Reproducibility (trying to take random out of random)

In short how a neural network learns:

`start with random numbers -> tensor operations -> update random numbers to try and make them better representations of the data -> again -> again -> again...`

To reduce the randomness in neural networks and PyTorch comes the concept of a `random seed`.

Essentially what the random seed does is 'flavour' the randomness

```python
import torch

# Create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducible tensors
import torch

# Set the random seed
RANDOM_SEED = 42 # can be anything

torch.manual_seed(RANDOM_SEED) # works only for a single block of code
random_tensor_C  = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D  = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
